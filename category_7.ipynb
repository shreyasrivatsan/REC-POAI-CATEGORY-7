```python
colors = ['Red', 'Green', 'Blue']

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

color = {}

def safe(node, c):
    for neighbor in graph[node]:
        if color.get(neighbor) == c:
            return False
    return True

def solve(node):
    if node == len(graph):
        return True

    region = list(graph.keys())[node]

    for c in colors:
        if safe(region, c):
            color[region] = c

            if solve(node + 1):
                return True

            del color[region]

    return False

if solve(0):
    print("Map Coloring Solution:")
    for region in color:
        print(region, ":", color[region])
else:
    print("No solution exists")
```

```python
from itertools import permutations

letters = 'SENDMORY'

for p in permutations(range(10), 8):
    d = dict(zip(letters, p))

    if d['S'] == 0 or d['M'] == 0:
        continue

    SEND = (d['S'] * 1000 + d['E'] * 100 +
            d['N'] * 10 + d['D'])

    MORE = (d['M'] * 1000 + d['O'] * 100 +
            d['R'] * 10 + d['E'])

    MONEY = (d['M'] * 10000 + d['O'] * 1000 +
             d['N'] * 100 + d['E'] * 10 + d['Y'])

    if SEND + MORE == MONEY:
        print("SEND =", SEND)
        print("MORE =", MORE)
        print("MONEY =", MONEY)
        break
```

```python
import random

n = 4
board = list(range(n))
random.shuffle(board)

def conflicts(board):
    count = 0

    for i in range(n):
        for j in range(i + 1, n):
            if abs(board[i] - board[j]) == abs(i - j):
                count += 1

    return count

while conflicts(board) > 0:
    best = conflicts(board)
    new_board = board[:]

    for i in range(n):
        for row in range(n):
            temp = board[:]
            temp[i] = row

            if conflicts(temp) < best:
                best = conflicts(temp)
                new_board = temp

    if new_board == board:
        random.shuffle(board)
    else:
        board = new_board

print("Solution:", board)
print("Conflicts:", conflicts(board))
```
